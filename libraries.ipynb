```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y = [5,8,6,10]
plt.scatter(x,y)
plt.title("My First Graph")
plt.xlabel("Age")
plt.ylabel("students")
plt.show()
```

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y = [5,8,6,10]
plt.bar(x,y)
plt.title("My First Graph")
plt.xlabel("Age")
plt.ylabel("students")
plt.show()
```

```python
import matplotlib.pyplot as plt

subjects = ["Python", "Java", "C", "JavaScript"]
students = [40, 25, 20, 15]

plt.pie(students, labels=subjects, autopct="%1.1f%%")

plt.title("Students Learning Programming Languages")

plt.show()
```

```python
import matplotlib.pyplot as plt

marks = [45, 50, 55, 60, 65, 70, 72, 75, 80, 82,
         85, 88, 90, 92, 95, 78, 67, 59, 73, 81]

plt.hist(marks)

plt.title("Histogram of Student Marks")
plt.xlabel("Marks")
plt.ylabel("Frequency")

plt.show()
```
